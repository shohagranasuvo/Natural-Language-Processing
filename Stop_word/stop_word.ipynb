```python
nlp=spacy.load("en_core_web_sm")
```

```python
doc=nlp("Hi I'm here for you .Can I give a suggestion about your next step ?")
```

```python
for token in doc :
    if not token.is_stop:
        print(token)
```

```python
not_stop_word =[]
```

```python
not_stop_word.clear()
for token in doc :
    
    if not token.is_stop and not token.is_punct :
            not_stop_word.append(token)
```

```python
print(not_stop_word)
```
