```python
from sklearn.feature_extraction.text import CountVectorizer
```

```python
v=CountVectorizer(ngram_range=(1,3))
```

```python
arr =['Great things comes with a great scarifies' , 'That\'s is okay' ,'Be brave and go on an adventure', 'what you do today can improve all your tomorows']
```

```python
v.fit(arr)
```

```python
import spacy
```

```python
nlp =spacy.load('en_core_web_sm')
```

```python
def preprocessing(arr) :
    text=""
    text=" ".join(arr)
    doc =nlp(text)
    filtered_token=[]
    for token in doc :
        
        if token.is_stop and token.is_punct :
            continue
        else : 
            filtered_token.append(token.lemma_)
    final =" ".join(filtered_token)     
    return final
```

```python
text =preprocessing(arr)
```

```python
print(text)
```

```python
v.fit([text])
```

```python
v.vocabulary_
```

```python
v.transform(["Great things comes with a great scarifies"]).toarray()
```
