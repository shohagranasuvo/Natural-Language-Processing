```python
import tensorflow_text as text
```

```python
preprocess_url = "https://tfhub.dev/tensorflow/bert_en_uncased_preprocess/3"
encoder_url = "https://tfhub.dev/tensorflow/bert_en_uncased_L-12_H-768_A-12/4"
```

```python
model_preprocess =tf.KerasLayer(preprocess_url)
```

```python
text = ["what you do Today can improve all your tomorrows" ,"I was born to be Awesome not perfect"]
```

```python
reselt_preprocess =model_preprocess(text)
```

```python
reselt_preprocess.keys()
```

```python
reselt_preprocess["input_type_ids"]
```

```python
reselt_preprocess["input_word_ids"]
```

```python
reselt_preprocess["input_mask"]
```

```python
model_encode =tf.KerasLayer(encoder_url)
```

```python
result_encode =model_encode(reselt_preprocess)
```
